```python
import requests
from bs4 import BeautifulSoup
import time
import pandas as pd
```

```python
import pandas as pd
```

```python
def collect_bbc_somali_article_links(base_url, start_page=1, end_page=1):
    headers = {
        "User-Agent": "Mozilla/5.0"
    }
    
    links = set()
    
    for page_num in range(start_page, end_page + 1):
        url = f"{base_url}?page={page_num}"
        
        try:
            response = requests.get(url, headers=headers, timeout=20)
            response.raise_for_status()
            
            soup = BeautifulSoup(response.text, "html.parser")
            
            article_anchors = soup.select('a[href*="/somali/articles/"]')
            
            for a in article_anchors:
                href = a.get("href")
                if href and href.startswith("https://www.bbc.com/somali/articles/"):
                    links.add(href)
            
            print(f"Page {page_num}: collected {len(links)} total links so far")
            time.sleep(3)
        
        except Exception as e:
            print(f"Error on page {page_num}: {e}")
    
    return list(links)
```

```python



base_url = "https://www.bbc.com/somali/topics/cwr9jry737xt"

links = sorted(collect_bbc_somali_article_links(base_url, start_page=1, end_page=24))

print(f"\nTotal links found: {len(links)}")
for link in links[:20]:
    print(link)
    
df_links = pd.DataFrame({"url": links})
df_links.to_csv("bbc_somali_siyaasad.csv", index=False, encoding="utf-8-sig")

print(f"Saved {len(df_links)} links")
```
