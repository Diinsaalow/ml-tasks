```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time
import os
```

```python
def scrape_somali_news(url):
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
    }
    try:
        response = requests.get(url, headers=headers, timeout=15)
        if response.status_code == 200:
            soup = BeautifulSoup(response.content, 'html.parser')
            
            # 1. Get the Headline
            headline = soup.find('h1').get_text(strip=True) if soup.find('h1') else "No Headline Found"
            
            # 2. Get the Body
            # Goobjoog specifically uses 'entry-content' or 'td-post-content'
            article_div = soup.find('div', class_='entry-content') or \
                          soup.find('div', class_='td-post-content') or \
                          soup.find('div', class_='article-content')
            
            if article_div:
                paragraphs = article_div.find_all('p')
                body_content = "\n".join([p.get_text(strip=True) for p in paragraphs if p.get_text(strip=True)])
            else:
                # Fallback
                paragraphs = soup.find_all('p')
                body_content = "\n".join([p.get_text(strip=True) for p in paragraphs if p.get_text(strip=True)])

            return headline, body_content
        else:
            return None, f"Error: {response.status_code}"
    except Exception as e:
        return None, f"Failed: {str(e)}"

def run_full_scrape(input_file="laacibnet_all_categories_links.xlsx", output_file="laacibnet_dataset.xlsx"):
    # Load the links
    if not os.path.exists(input_file):
        print(f"❌ File-ka {input_file} lama helin!")
        return

    df = pd.read_excel(input_file)
    total_links = len(df)
    print(f"📄 Waxaa la helay {total_links} xiriiriye. Bilaabaya scraping-ka...")

    # Prepare columns
    headlines = []
    bodies = []

    for index, row in df.iterrows():
        url = row['link']
        print(f"🌐 [{index + 1}/{total_links}] Scrapping: {url}")
        
        title, content = scrape_somali_news(url)
        headlines.append(title)
        bodies.append(content)

        # Checkpoint: Save every 20 articles
        if (index + 1) % 20 == 0:
            temp_df = df.iloc[:len(headlines)].copy()
            temp_df['Headline'] = headlines
            temp_df['Body'] = bodies
            temp_df.to_excel(output_file, index=False)
            print(f"💾 Checkpoint: {index + 1} maqaal ayaa la keydiyay.")

        # Respectful delay
        time.sleep(0.5)

    # Final Save
    df['Headline'] = headlines
    df['Body'] = bodies
    df.to_excel(output_file, index=False)
    print(f"\n✅ DHAMAAD! Xogtii oo dhan waxay ku jirtaa: {output_file}")

if __name__ == "__main__":
    run_full_scrape()
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import os
import time

def collect_links_intelligent_resume(categories_dict, output_file, max_stagnation=5):
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
    }

    # 1. Load the existing Excel data
    if os.path.exists(output_file):
        df_master = pd.read_excel(output_file)
        all_links = df_master.to_dict('records')
        existing_urls = set(df_master['link'].tolist())
    else:
        all_links = []
        existing_urls = set()
        df_master = pd.DataFrame()

    for category_name, base_url in categories_dict.items():
        # 2. Identify the last link saved for THIS category
        last_link_in_category = None
        if not df_master.empty:
            cat_df = df_master[df_master['category'] == category_name]
            if not cat_df.empty:
                last_link_in_category = cat_df['link'].iloc[-1]

        print(f"\n--- Category: {category_name} ---")
        if last_link_in_category:
            print(f"Targeting Resume Point: {last_link_in_category}")
        
        page = 1
        stagnation_counter = 0
        found_resume_point = False if last_link_in_category else True

        while True:
            url = f"{base_url}page/{page}/"
            try:
                response = requests.get(url, headers=headers, timeout=12)
                
                if response.status_code != 200:
                    print(f"End of category reached (Status {response.status_code}).")
                    break

                soup = BeautifulSoup(response.content, 'html.parser')
                post_links = soup.find_all('a', rel='bookmark')

                if not post_links:
                    break

                new_on_page = 0
                for link in post_links:
                    href = link.get('href')
                    
                    # If we haven't found the resume point yet, skip until we do
                    if not found_resume_point:
                        if href == last_link_in_category:
                            found_resume_point = True
                            print(f"Resume point found on Page {page}. Continuing...")
                        continue

                    if href and href not in existing_urls:
                        all_links.append({
                            'title': link.get('title'),
                            'link': href,
                            'category': category_name,
                            'scraped_at': time.strftime("%Y-%m-%d %H:%M:%S")
                        })
                        existing_urls.add(href)
                        new_on_page += 1

                # If we searched many pages and didn't find the resume point, 
                # it might be very old. We should start collecting again.
                if not found_resume_point and page > 20: 
                     print("Resume point not found in last 20 pages. Starting from Page 1.")
                     found_resume_point = True

                # Logic to handle moving forward
                if found_resume_point:
                    if new_on_page == 0:
                        stagnation_counter += 1
                    else:
                        stagnation_counter = 0
                    
                    if stagnation_counter >= max_stagnation:
                        print("No new links found. Category likely up to date.")
                        break

                    # Save progress to Excel after every page
                    pd.DataFrame(all_links).to_excel(output_file, index=False)
                    print(f"Page {page} processed. {new_on_page} new links added.")
                
                page += 1
                time.sleep(1.2)

            except (requests.ConnectionError, requests.Timeout):
                print(f"Network issue on Page {page}. Retrying in 15s...")
                time.sleep(15)
                continue
            except KeyboardInterrupt:
                print("\nManual stop detected. Data is saved in Excel.")
                return

    print("\nExtraction finished and saved to Excel.")

# --- RUN CONFIG ---
categories = {
    "ciyaaraha_maanta": "https://www.laacibnet.net/category/wararka-ciyaaraha-maanta/",
    "premier_league": "https://www.laacibnet.net/category/wararka-premier-league/",
    "la_liga": "https://www.laacibnet.net/category/wararka-la-liga/"
}

if __name__ == "__main__":
    collect_links_intelligent_resume(categories, "laacibnet.xlsx")
```
